1.Download a CSV file containing a list of movies with columns like 'title', 'genre', 'rating', and 'box_office', then use pandas to load this file and select only the 'title' and 'rating' columns using the loc method.

2.Given a DataFrame of Flipkart products with columns 'product_name', 'category', and 'price', use iloc to select the first 10 rows and all columns, and print the result.


```python
import pandas as pd

df = pd.DataFrame({
    "product_name": ["Laptop", "Mobile", "Headphones", "Watch", "Keyboard",
                     "Mouse", "Monitor", "Tablet", "Speaker", "Camera"],
    "category": ["Electronics", "Mobile", "Audio", "Wearable", "Computer",
                 "Computer", "Electronics", "Tablet", "Audio", "Camera"],
    "price": [55000, 25000, 1500, 3000, 1200,
              700, 15000, 18000, 2500, 35000]
})

# Select first 10 rows and all columns
result = df.iloc[:10, :]

print(result)
```

3.Using a DataFrame of food orders from Zomato with columns 'restaurant', 'order_amount', and 'delivery_time', filter and display only those orders where 'order_amount' is greater than 500 using a boolean condition.


```python
import pandas as pd

df = pd.DataFrame({
    "restaurant": ["Spice Hub", "Pizza Point", "Burger House",
                   "Biryani Zone", "Cafe Aroma"],
    "order_amount": [450, 750, 320, 650, 900],
    "delivery_time": [30, 45, 25, 50, 35]
})

# Filter orders greater than 500
result = df[df["order_amount"] > 500]

print(result)
```

4.Given a DataFrame of Spotify songs with columns 'song', 'artist', 'streams', and 'duration', use the query() method to select all songs with more than 1,000,000 streams and duration less than 180 seconds.<br><br><em><strong>Hint:</strong> Use the query syntax: query('streams > 1000000 and duration < 180')</em>


```python
import pandas as pd

df = pd.DataFrame({
    "song": ["Song A", "Song B", "Song C", "Song D", "Song E"],
    "artist": ["Artist 1", "Artist 2", "Artist 3", "Artist 4", "Artist 5"],
    "streams": [1500000, 800000, 2200000, 1200000, 500000],
    "duration": [170, 200, 175, 165, 150]
})

# Filter songs with streams > 1,000,000 and duration < 180 seconds
result = df.query("streams > 1000000 and duration < 180")

print(result)
```

5.Suppose you have a DataFrame of IPL cricket matches with columns 'team1', 'team2', 'winner', and 'total_runs'. Write code to filter and display all matches where 'total_runs' is between 180 and 220 using boolean indexing.<br><br><em><strong>Constraint:</strong> Do not use the query() method for this task.</em>


```python
import pandas as pd

df = pd.DataFrame({
    "team1": ["CSK", "MI", "RCB", "KKR", "RR"],
    "team2": ["MI", "RCB", "KKR", "RR", "DC"],
    "winner": ["CSK", "MI", "RCB", "RR", "DC"],
    "total_runs": [195, 175, 210, 225, 185]
})

# Filter total runs between 180 and 220
result = df[(df["total_runs"] >= 180) & (df["total_runs"] <= 220)]

print(result)
```
